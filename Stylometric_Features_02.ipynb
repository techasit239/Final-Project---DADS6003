```python
import os
import pandas as pd
import numpy as np
import IPython
from IPython.display import display
import gdown
```

```python
train_file_id = "1NZIkHAGvoqcMOFAj_ONpi-0GRc6ZNEwN"
test_file_id = "1NJcC72cNH0PbR4hRq1OpZvUF5OHzU1ye"
train_url = f"https://drive.google.com/uc?id={train_file_id}"
test_url = f"https://drive.google.com/uc?id={test_file_id}"
```

```python
output_train = "train.csv"
output_test = "test.csv"
gdown.download(train_url, output_train, quiet=False)
gdown.download(test_url, output_test, quiet=False)
```

```python
df = pd.read_csv(output_train)
df.head()
```

```python
# Punctuation Features (2 features)
def punctuation_features(row):
  s = str(row["Subject"]) + " " + str(row["Body"])
  s_count = len(s)
  target_punc = ['.', ',', ':', '-', '"', '(', ')', '/', '\\' ]

  total_punc_count = 0
  variety_count = 0

  if (s_count==0):
    return {
        "punctuation_frequency" : 0.0,
        "punctuation_variety" : 0
    }
  for punc in target_punc:
    punc_count = s.count(punc)
    total_punc_count += punc_count
    if (punc_count>0):
      variety_count += 1
  return {
      "punctuation_frequenc": total_punc_count/s_count,
      "punctutation_variety" : variety_count
  }

feature_result = df.apply(punctuation_features, axis=1, result_type="expand")
df_punc = pd.concat([df, feature_result], axis=1)
display(df_punc.head())

output_fname = "train_processed.csv"
df_punc.to_csv(output_fname, index = False)
```

```python
# Readability Scores Features (5 features)
## 1. Flesch Reading Ease
## 2. SMOG Index
## 3. Dale--Chall Readability Score
## 4. Coleman--Liau Index (CLI)
## 5. Gunning Fog Index (GFI)

def readability_score_features(row):
  s = str(row["Subject"]) + " " + str(row["Body"])
  fre = textstat.flesch_reading_ease(s)
  smog = textstat.smog_index(s)
  dale_chall = textstat.dale_chall_readability_score_v2(s)
  cli = textstat.coleman_liau_index(s)
  return {
      "fleasch_reading_ease" : fre,
      "smog_index": smog,
      "dale_chall_readability_score": dale_chall,
      "coleman_Liau_index": cli,
      "gunning_fog_index": textstat.gunning_fog(s)
  }

feature_result = df.apply(readability_score_features, axis=1, result_type="expand")
feature_result.head()

df_features = pd.concat([df_punc, feature_result], axis=1)
display(df_features.head())

df_features.to_csv("train_processed.csv", index=False)
```
